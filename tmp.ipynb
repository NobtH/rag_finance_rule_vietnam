```python
import pandas as pd

# Bước 1: Đọc tệp CSV vào DataFrame
try:
    df = pd.read_csv('data/Chatbot - Data CSCN (2).csv')
except FileNotFoundError:
    print("Lỗi: Không tìm thấy tệp 'data.csv'.")
    exit()

# Bước 2: Sửa đổi cột 'Value'
# Sử dụng .str.replace() để thay thế tất cả các ký tự '#' bằng một chuỗi rỗng
df['Context'] = df['Context'].str.replace('# ', '', regex=False)


df.to_csv('data/Chatbot - Data CSCN (2).csv', index=False)

print("Đã sửa đổi và lưu tệp thành công thành 'data_modified.csv'")
print("\nNội dung mới của tệp:")
print(df)
```
